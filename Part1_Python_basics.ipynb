### 🕸️ Network calculations from the lectures

Now we can calculate some measures from Weeks 2 and 3 for the small network on the slides (4 nodes, 4 links).

**Maximum possible number of links** in an undirected network: $L_{max} = \dfrac{N(N-1)}{2}$

````python
L_max = N * (N - 1) / 2
L_max
````

**Density**: the actual number of links divided by the possible number of links, $\dfrac{L}{L_{max}}$

````python
density = L / L_max
density
````

**Average degree** of an undirected network: $\langle k \rangle = \dfrac{2L}{N}$ (every link is counted by both of its nodes, hence the 2).

````python
average_degree = 2 * L / N
average_degree
````

`print()` lets us show several results with a label. Put the label in quotes and separate the parts with commas:

````python
print("Number of nodes:", N)
print("Number of links:", L)
print("Density:", density)
print("Average degree:", average_degree)
````

## 2. Lists

A **list** stores several values in order. Lists are written with square brackets `[ ]` and commas between items. Text (a **string**) is always written in quotes.

````python
friends_of_ana = ["Ben", "Chen", "Dara"]
friends_of_ana
````

### Getting items out of a list

Use square brackets with a position number to get one item.

⚠️ **Careful: counting in Python starts at `0`!** The first item is `[0]`, the second is `[1]`, and so on.

````python
friends_of_ana[0]
````

````python
friends_of_ana[1]
````

⚠️ **Error on purpose.** The list has 3 items, so the positions are 0, 1 and 2. What happens if we ask for position 3?

````python
friends_of_ana[3]   # to get the third item you need friends_of_ana[2]
````

Read the last line of the error message: `IndexError: list index out of range`. Error messages usually tell you exactly what went wrong.

Negative numbers count from the end, and `[i:j]` gives a **slice** from position `i` up to (but not including) `j`:

````python
print(friends_of_ana[-1])    # last item
print(friends_of_ana[0:2])   # items at positions 0 and 1
print(friends_of_ana[1:])    # from position 1 to the end
````

### Useful things to do with lists

````python
print(len(friends_of_ana))        # how many items? (len = length)
print("Chen" in friends_of_ana)   # is "Chen" in the list?
print("Eli" in friends_of_ana)
````

🕸️ `len(friends_of_ana)` is the number of Ana's friends: her **degree**.

Lists can grow. `append()` adds **one** item to the end; `extend()` adds **all items of another list**:

````python
friends_of_ana.append("Eva")
friends_of_ana
````

````python
friends_of_ana.extend(["Felix", "Gus"])
friends_of_ana
````

Notice the difference: if you `append` a list, the whole list becomes **one single item** inside your list.

````python
test_list = ["Ben", "Chen"]
test_list.append(["Dara", "Eva"])
test_list   # the last item is itself a list!
````

### Lists of numbers

Lists can also hold numbers. Here is the degree of each person in a small friendship network. Python has handy tools for such lists:

````python
degrees = [4, 3, 4, 5, 2, 4, 6, 4, 3, 2, 3, 2]

print("Number of people:", len(degrees))
print("Sum of degrees:", sum(degrees))
print("Highest degree:", max(degrees))
print("Lowest degree:", min(degrees))
print("Sorted:", sorted(degrees))
````

🕸️ The average degree is the sum of all degrees divided by the number of nodes:

````python
sum(degrees) / len(degrees)
````

## 3. Tuples 🕸️

A **tuple** is like a list, but written with round brackets `( )` and it **cannot be changed** after it is created.

Why do we care? Because a link between two nodes is a pair — and as on the Week 2 slides, we write each link as a tuple.

````python
link = (1, 2)
print(link[0])   # first node
print(link[1])   # second node
````

You can **unpack** a tuple into two variables in one line:

````python
node_a, node_b = link
print(node_a)
print(node_b)
````

⚠️ **Error on purpose.** Tuples cannot be modified:

````python
link[0] = 3
````

### 🕸️ A list of tuples is an edge list

On the Week 2 slides, the network with 4 nodes was written as `{(1, 2), (1, 4), (3, 4), (2, 4)}`. In Python we write it as a **list of tuples**:

````python
edges = [(1, 2), (1, 4), (3, 4), (2, 4)]

print("The first link is:", edges[0])
print("Number of links L:", len(edges))
````

Nodes can also be names. Here is the friendship network we will use in Part 2:

````python
friendships = [
    ("Ana", "Ben"), ("Ana", "Chen"), ("Ana", "Dara"), ("Ana", "Eva"),
    ("Ben", "Chen"), ("Ben", "Dara"), ("Chen", "Dara"), ("Dara", "Eva"),
    ("Dara", "Felix"), ("Chen", "Felix"), ("Felix", "Gus"), ("Felix", "Hana"),
    ("Gus", "Hana"), ("Gus", "Ivo"), ("Gus", "Jonas"), ("Gus", "Kim"),
    ("Gus", "Lena"), ("Hana", "Ivo"), ("Kim", "Lena"), ("Ivo", "Jonas"),
    ("Hana", "Kim"),
]

L = len(friendships)
N = 12   # Ana, Ben, Chen, Dara, Eva, Felix, Gus, Hana, Ivo, Jonas, Kim, Lena

print("Links:", L)
print("Density:", L / (N * (N - 1) / 2))
````

## 4. pandas 🐼: tables of data

Real network data usually comes as a spreadsheet, for example a CSV file with one row per link. The `pandas` library lets us work with such tables in Python. A table in pandas is called a **DataFrame** — think of it as a spreadsheet inside Python.

First we **import** the library. `as pd` gives it a short nickname, so we can write `pd` instead of `pandas`.

````python
import pandas as pd
````

### An edge list as a table

We can turn our list of tuples straight into a DataFrame. Each tuple becomes a row; we name the two columns `source` and `target`.

````python
edges_df = pd.DataFrame(friendships, columns=["source", "target"])
edges_df
````

Some useful ways to inspect a table:

````python
edges_df.head()   # the first 5 rows
````

````python
edges_df.shape    # (number of rows, number of columns)
````

````python
edges_df.columns  # the column names
````

````python
edges_df["source"]   # select one column with its name in square brackets
````

🕸️ In an edge list, **each row is one link**, so the number of rows is `L`.

### A node table

We can also make a table with **one row per node** and information about each node in the columns. Again we start from a list of tuples: (name, group, degree).

````python
nodes = [
    ("Ana", "left", 4), ("Ben", "left", 3), ("Chen", "left", 4),
    ("Dara", "left", 5), ("Eva", "left", 2), ("Felix", "middle", 4),
    ("Gus", "right", 6), ("Hana", "right", 4), ("Ivo", "right", 3),
    ("Jonas", "right", 2), ("Kim", "right", 3), ("Lena", "right", 2),
]

nodes_df = pd.DataFrame(nodes, columns=["name", "group", "degree"])
nodes_df
````

````python
print("Average degree:", nodes_df["degree"].mean())
print("Highest degree:", nodes_df["degree"].max())
````

````python
nodes_df[nodes_df["degree"] >= 4]
````
